**(4) Write a python script to take an input angle in degrees and print the values of all the 6
trigonometric functions for this angle. Remember to write a line of code to convert your
input into radians before finding the trigonometric values.**

```python
import math

# Input angle in degrees
angle = float(input("Enter an angle in degrees: "))

# Convert to radians
radian = math.radians(angle)

# Trigonometric values
sin_angle = math.sin(radian)
cos_angle = math.cos(radian)
tan_angle = math.tan(radian)

# Tolerance for floating-point comparisons
eps = 1e-10

# Fix very small values
if abs(sin_angle) < eps:
    sin_angle = 0.0

if abs(cos_angle) < eps:
    cos_angle = 0.0

if abs(tan_angle) < eps:
    tan_angle = 0.0

# sec
if abs(cos_angle) < eps:
    sec_angle = "Undefined"
else:
    sec_angle = 1 / cos_angle

# cosec
if abs(sin_angle) < eps:
    cosec_angle = "Undefined"
else:
    cosec_angle = 1 / sin_angle

# cot
if abs(tan_angle) < eps:
    cot_angle = "Undefined"
else:
    cot_angle = 1 / tan_angle

# Output
print(f"sin({angle}) = {sin_angle}")
print(f"cos({angle}) = {cos_angle}")
print(f"tan({angle}) = {tan_angle}")
print(f"sec({angle}) = {sec_angle}")
print(f"cosec({angle}) = {cosec_angle}")
print(f"cot({angle}) = {cot_angle}")
```

**(5) Write a program to find the roots of a quadratic equation. Figure out what the inputs
should be and also have provision to deal with cases where you end up getting complex
roots.**

```python
import cmath

# Quadratic equation is ax^2 + bx + c = 0

a = float(input("Enter the value of a: "))
b = float(input("Enter the value of b: "))
c = float(input("Enter the value of c: "))

# Calculate the discriminant
d = (b**2) - (4*a*c)

print(f"The discriminant (d) is: {d}")

# Find roots based on the discriminant
if d > 0:
    # Real and distinct roots
    x1 = (-b + cmath.sqrt(d)) / (2 * a)
    x2 = (-b - cmath.sqrt(d)) / (2 * a)
    print(f"Two distinct real roots: x1 = {x1:.2f}, x2 = {x2:.2f}")
elif d == 0:
    # Real and equal roots
    x = -b / (2 * a)
    print(f"One real root (repeated): x = {x:.2f}")
else:
    # Complex roots
    x1 = (-b + cmath.sqrt(d)) / (2 * a)
    x2 = (-b - cmath.sqrt(d)) / (2 * a)
    print(f"Two complex roots: x1 = {x1:.2f}, x2 = {x2:.2f}")
```

**(6) Write a program to print the sum of the digits in a number.**

```python
# Program to find the sum of digits of a number

num = int(input("Enter a number: "))

total = 0

while num > 0:
    digit = num % 10
    total = total + digit
    num = num // 10

print("Sum of digits =", total)
```

**(7) Write a script to check if a number is palindrome or not.**

```python
num = int(input("Enter the number "))
original = num
reverse = 0

while num > 0 :
  digits = num % 10
  reverse = reverse * 10 + digits
  num = num // 10

if original == reverse :
  print("Its Palindrome")
else :
  print("Its not a Palindrome")
```
